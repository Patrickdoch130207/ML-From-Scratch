```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder, OneHotEncoder
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.model_selection import train_test_split
from k_means import KMeans
from sklearn.metrics import silhouette_score
```

**Exploration des données,vérification des valeurs manquantes,dupliquées,etc**

```python
data = pd.read_csv("Mall_Customers.csv")
data.head()
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
data.duplicated().sum()
```

**Encodage des variables catégorielles**

```python
le = LabelEncoder()
data['Gender'] = le.fit_transform(data['Gender'])
data
```

**Mise à l'échelle des variables numériques**

```python
scaler = StandardScaler()
data = data.drop('CustomerID', axis=1)
X = scaler.fit_transform(data)
X
```

**model1 : methode 'forgy',k=5**

```python
model1 = KMeans(k=5,random_state=42)
```

```python
model1.fit(X)
```

```python
print(model1.calculate_silhouette_score(X))
print(silhouette_score(X,model1.labels))
```

**model2 : methode 'random_partition',k=2**

```python
model2 = KMeans(k=5,init_method = 'random_partition',random_state=42)
model2.fit(X)
```

```python
print(model2.calculate_silhouette_score(X))
print(silhouette_score(X,model2.labels))
```

**model3 : methode 'kmeans++',k=2**

```python
model3 = KMeans(k=5,init_method = 'kmeans++',random_state=42)
model3.fit(X)
```

```python
print(model3.calculate_silhouette_score(X))
print(silhouette_score(X,model3.labels))
```

```python
print(f"Forgy          : {model1.calculate_silhouette_score(X):.4f}")
print(f"Random partition: {model2.calculate_silhouette_score(X):.4f}")
print(f"KMeans++       : {model3.calculate_silhouette_score(X):.4f}")
```
